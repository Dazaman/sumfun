# Working with Variables and Data Types
Learn about integers, floats, strings, and booleans. Practice declaring variables and understanding variable assignment.

```python
# Working with Variables and Data Types

# Integers
a = 10  # Assigning an integer value to variable 'a'
b = 5   # Assigning another integer value to variable 'b'

# Floats
pi = 3.14  # Assigning a float value to variable 'pi'
gravity = 9.81  # Assigning another float value to variable 'gravity'

# Strings
greeting = "Hello, World!"  # Assigning a string value to variable 'greeting'
name = "Alice"  # Assigning another string value to variable 'name'

# Booleans
is_sunny = True  # Assigning a boolean value to variable 'is_sunny'
is_raining = False  # Assigning another boolean value to variable 'is_raining'

# Printing variables
print(a)
print(pi)
print(greeting)
print(is_sunny)

# Variable reassignment
a = 20  # Reassigning a new integer value to variable 'a'
greeting = "Hi, there!"  # Reassigning a new string value to variable 'greeting'

# Printing reassigned variables
print(a)
print(greeting)
```

# Basic Operations and Math
Explore arithmetic operators (+, -, *, /), comparison operators (>, <, ==), and logical operators (and, or, not).

```python
# Basic Operations and Math

# Arithmetic Operators
sum_result = a + b  # Addition
difference_result = a - b  # Subtraction
product_result = a * b  # Multiplication
quotient_result = a / b  # Division

# Printing arithmetic results
print("Sum:", sum_result)
print("Difference:", difference_result)
print("Product:", product_result)
print("Quotient:", quotient_result)

# Comparison Operators
is_equal = a == b  # Equality
is_not_equal = a != b  # Inequality
is_greater = a > b  # Greater than
is_less = a < b  # Less than
is_greater_or_equal = a >= b  # Greater than or equal to
is_less_or_equal = a <= b  # Less than or equal to

# Printing comparison results
print("Is Equal:", is_equal)
print("Is Not Equal:", is_not_equal)
print("Is Greater:", is_greater)
print("Is Less:", is_less)
print("Is Greater or Equal:", is_greater_or_equal)
print("Is Less or Equal:", is_less_or_equal)

# Logical Operators
and_result = is_sunny and is_raining  # Logical AND
or_result = is_sunny or is_raining  # Logical OR
not_result = not is_sunny  # Logical NOT

# Printing logical results
print("AND Result:", and_result)
print("OR Result:", or_result)
print("NOT Result:", not_result)
```

# Control Flow with If Statements
Learn how to make decisions in code using if, elif, and else statements with practical examples.

```python
# Control Flow with If Statements

# Using if statement to check a condition
if a > b:
    print("a is greater than b")

# Using if-else statement to check a condition and provide an alternative
if is_sunny:
    print("It's a sunny day!")
else:
    print("It's not a sunny day.")

# Using if-elif-else statement to check multiple conditions
if a == b:
    print("a is equal to b")
elif a > b:
    print("a is greater than b")
else:
    print("a is less than b")

# Nested if statements
if is_sunny:
    if is_raining:
        print("It's sunny and raining at the same time!")
    else:
        print("It's sunny but not raining.")
else:
    print("It's not sunny.")

# Using logical operators with if statements
if is_sunny and not is_raining:
    print("It's sunny and not raining.")
elif is_sunny and is_raining:
    print("It's sunny and raining.")
elif not is_sunny and is_raining:
    print("It's not sunny but it's raining.")
else:
    print("It's neither sunny nor raining.")
```

# Working with Loops
Understand how to use for and while loops to repeat actions, including loop control statements like break and continue.

```python
# Working with Loops

# Using a for loop to iterate over a range of numbers
for i in range(5):
    print("Iteration:", i)

# Using a for loop to iterate over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print("Fruit:", fruit)

# Using a while loop to repeat actions until a condition is met
count = 0
while count < 5:
    print("Count:", count)
    count += 1  # Incrementing the count

# Using break statement to exit a loop prematurely
for i in range(10):
    if i == 5:
        break  # Exit the loop when i is 5
    print("Break Loop Iteration:", i)

# Using continue statement to skip the current iteration and continue with the next one
for i in range(10):
    if i % 2 == 0:
        continue  # Skip the iteration if i is even
    print("Continue Loop Iteration:", i)
```

# Creating and Using Functions
Learn how to define and call functions, pass arguments, and return values from functions.

```python
# Creating and Using Functions

# Defining a simple function
def greet():
    print("Hello, welcome to the Python tutorial!")

# Calling the function
greet()

# Defining a function with parameters
def greet_person(name):
    print(f"Hello, {name}! Welcome to the Python tutorial!")

# Calling the function with an argument
greet_person("Alice")

# Defining a function that returns a value
def add_numbers(x, y):
    return x + y

# Calling the function and storing the result in a variable
result = add_numbers(10, 5)
print("Sum:", result)

# Defining a function with default parameters
def greet_with_default(name="Guest"):
    print(f"Hello, {name}! Welcome to the Python tutorial!")

# Calling the function without an argument
greet_with_default()

# Calling the function with an argument
greet_with_default("Bob")

# Defining a function with multiple return values
def arithmetic_operations(x, y):
    sum_result = x + y
    difference_result = x - y
    product_result = x * y
    quotient_result = x / y
    return sum_result, difference_result, product_result, quotient_result

# Calling the function and unpacking the returned values
sum_result, difference_result, product_result, quotient_result = arithmetic_operations(20, 5)
print("Sum:", sum_result)
print("Difference:", difference_result)
print("Product:", product_result)
print("Quotient:", quotient_result)
```

# Lists and List Operations
Work with lists, including creating, accessing, modifying, and common list methods like append() and remove().

```python
# Lists and List Operations

# Creating a list
fruits = ["apple", "banana", "cherry"]
print("Fruits:", fruits)

# Accessing list elements by index
print("First fruit:", fruits[0])
print("Second fruit:", fruits[1])
print("Third fruit:", fruits[2])

# Modifying list elements
fruits[1] = "blueberry"
print("Modified Fruits:", fruits)

# Adding elements to the list using append()
fruits.append("orange")
print("Fruits after append:", fruits)

# Removing elements from the list using remove()
fruits.remove("cherry")
print("Fruits after remove:", fruits)

# Using pop() to remove an element by index
popped_fruit = fruits.pop(1)
print("Popped Fruit:", popped_fruit)
print("Fruits after pop:", fruits)

# Using len() to get the length of the list
print("Number of fruits:", len(fruits))

# Iterating over a list using a for loop
for fruit in fruits:
    print("Fruit:", fruit)

# Checking if an element is in the list
if "apple" in fruits:
    print("Apple is in the list")

# Using list slicing
print("First two fruits:", fruits[:2])
print("Last fruit:", fruits[-1])

# Creating a list of numbers
numbers = [1, 2, 3, 4, 5]
print("Numbers:", numbers)

# Using list comprehension to create a new list
squares = [x**2 for x in numbers]
print("Squares:", squares)
```

# Working with Strings
Explore string operations, methods, and formatting including slicing and common string methods.

```python
# Working with Strings

# Creating strings
str1 = "Hello"
str2 = "World"

# Concatenating strings
concatenated_str = str1 + " " + str2
print("Concatenated String:", concatenated_str)

# String length
length = len(concatenated_str)
print("Length of Concatenated String:", length)

# Accessing characters in a string
first_char = concatenated_str[0]
last_char = concatenated_str[-1]
print("First Character:", first_char)
print("Last Character:", last_char)

# Slicing strings
substring = concatenated_str[0:5]
print("Substring (0-5):", substring)

# String methods
upper_str = concatenated_str.upper()
lower_str = concatenated_str.lower()
title_str = concatenated_str.title()
print("Uppercase String:", upper_str)
print("Lowercase String:", lower_str)
print("Title Case String:", title_str)

# Finding a substring
index = concatenated_str.find("World")
print("Index of 'World':", index)

# Replacing a substring
replaced_str = concatenated_str.replace("World", "Python")
print("Replaced String:", replaced_str)

# Splitting a string
split_str = concatenated_str.split(" ")
print("Split String:", split_str)

# Joining a list of strings
joined_str = " ".join(split_str)
print("Joined String:", joined_str)

# String formatting
name = "Alice"
age = 25
formatted_str = f"My name is {name} and I am {age} years old."
print("Formatted String:", formatted_str)

# Checking if a string starts or ends with a substring
starts_with_hello = concatenated_str.startswith("Hello")
ends_with_world = concatenated_str.endswith("World")
print("Starts with 'Hello':", starts_with_hello)
print("Ends with 'World':", ends_with_world)
```

# Basic Input and Output
Learn how to get user input using input() and display output using print(), including formatting options.

```python
# Basic Input and Output

# Getting user input using input() function
user_name = input("Enter your name: ")
user_age = input("Enter your age: ")

# Displaying output using print() function
print("Hello,", user_name)
print("You are", user_age, "years old")

# Formatting output using f-strings
print(f"Hello, {user_name}. You are {user_age} years old.")

# Formatting output using str.format() method
print("Hello, {}. You are {} years old.".format(user_name, user_age))

# Formatting output using % operator
print("Hello, %s. You are %s years old." % (user_name, user_age))
```
